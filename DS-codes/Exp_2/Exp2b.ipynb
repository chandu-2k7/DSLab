```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("student.db")
cursor=conn.cursor()
cursor.execute("CREATE TABLE IF NOT EXISTS students (id INTEGER,name TEXT)")
print("Table students craeted")
```

```python
cursor.execute("INSERT INTO students VALUES (1,'kani')")
print("Inserted successfully")
conn.commit()
```

```python
df=pd.read_sql_query("select *from students",conn)
print(df)
```

```python
#Update and delete
cursor.execute("UPDATE students SET name='kanishka' where id =1")
df=pd.read_sql_query("select *from students",conn)
print("Updated")
print("\n",df)
```

```python
cursor.execute("DELETE from students where id=1")
df=pd.read_sql_query("select *from students",conn)
print("After deleted")
print("\n",df)
```
