```python
# importing pandas
import pandas as pd
```

```python
# storing url of online csv file
url="https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
```

```python
# reading csv file
df=pd.read_csv(url)
```

```python
# Getting data from url csv file

print(df.head())
```

```python
#Storing all rows in csv file which named studentCSV.csv
df.to_csv("iris_output.csv",index=False)
print("Dataset of all rows successfully written to iris_output.csv")
```

```python
# storing first five rows in csv file which named student5csv.csv
temp=df.head()
temp.to_csv("student5csv.csv",index=False)
print("Dataset of 5 rows successfully written to student5csv.csv")
```
